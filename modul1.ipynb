# Геометрия и линейна алгебра
Този модул показва как геометрията и линейната алгебра помагат за дизайна на матрицата на дадено изделие. Когато пластмасите изтиват, те се свиват. Много е важно това да се вземе предвид, когато се проектира калъпа(матрицата) за дадено изделие. За да бъдат точни размерите на готовото изделие, матрицата трябва да е по-голяма от крайният продукт.
Приложените математически концепции са мащабиране и процентни съотношения.
Формула за за размер на матрицата ($D_m$):
$$ D_m = D_p \times(1+S)$$

($D_p$) - размер на детайла;
($S$) - коефициент на свиване на материала. Този коефициент е различен за различните материали.
В софтуерното инжинерство, вместо да се изчилява всеки един размер поотделно, използваме **линейна алгебра**. Представяме размерите на изделието като 3D вектор $\mathbf{v} = [L. W, H]^T$ и прилагаме диагонална матрица за мащабиране $\mathbf{T}$ :
$$\mathbf{v}_{matrix} = \mathbf{T} \cdot \mathbf{v} \implies \begin{bmatrix} L_m \\ W_m \\ H_m \end{bmatrix} = \begin{bmatrix} 1+S & 0 & 0 \\ 0 & 1+S & 0 \\ 0 & 0 & 1+S \end{bmatrix} \begin{bmatrix} L \\ W \\ H \end{bmatrix}$$


```python
import numpy as np

def calculate_mold_dimensions(dimensions_vector, shrinkage_rate):
    """
    Изчислява 3D размерите на матрицата чрез матрична трансформация.

    dimensions_vector - list или np.array с три елемента [Дължина, Ширина, Височина]
    shrinkage_rate - коефициент на свиване (float)
    return - np.array с мащабираните размери на матрицата
    """
    # 1. Дефиниране на 3D вектора на продукта
    v_product = np.array(dimensions_vector)

    # 2. Създаване на 3D скалираща матрица за трансформация (Identity matrix * factor)
    scale_factor = 1 + shrinkage_rate
    T_matrix = np.diag([scale_factor, scale_factor, scale_factor])

    # 3. Матрично умножение (Dot product)
    v_mold = np.dot(T_matrix, v_product)

    return np.round(v_mold, 2)

# Пример с реалните данни за "Wham Crystal Box 32L" от предходната таблица:
# Размери: L=595.0 mm, W=395.0 mm, H=180.0 mm
# Материал: Polypropylene със свиване S = 0.018 (1.8%)

product_name = "Wham Crystal Box 32L"
dimensions_3d = [595.0, 400.0, 180.0]
s_rate = 0.018

mold_dimensions = calculate_mold_dimensions(dimensions_3d, s_rate)

print(f"--- {product_name} ---")
print(f"Желани размери на продукта (L, W, H): {dimensions_3d} mm")
print(f"Коефициент на свиване (S): {s_rate}")
print(f"Изчислени 3D размери на кухината на матрицата: {mold_dimensions.tolist()} mm")

```
